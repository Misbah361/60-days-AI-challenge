step 1: setup

```python
import nltk, string
from collections import Counter
nltk.download('punkt'); nltk.download('punkt_tab'); nltk.download('stopwords')
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

text = """Machine learning is transforming the world. Companies use AI to search data,
power chatbots, and classify text. But machines don't read words; they read numbers!"""
print("ORIGINAL:\n", text)
```

step 2: Tokenization

```python
tokens = word_tokenize(text.lower())
print("BEFORE:", text[:60], "...")
print("AFTER (tokens):", tokens)
```

step 3: Removing punctuation and stopwords

```python
stop_words = set(stopwords.words('english'))
no_punct = [t for t in tokens if t not in string.punctuation]
clean = [t for t in no_punct if t not in stop_words]
print("Tokens:", len(tokens), "| after punctuation removal:", len(no_punct), "| after stop words:", len(clean))
print("CLEAN:", clean)
```

step 4: bag-of-words

```python
bow = Counter(clean)
print(bow)

vocab = sorted(bow)
vector = [bow[w] for w in vocab]
print("Vocabulary:", vocab)
print("Vector:    ", vector)
```

step 5: Building dataset

```python
import pandas as pd
docs = [
    "I love learning about artificial intelligence!",
    "Data science is fun and powerful.",
    "Chatbots use natural language processing.",
    "I love data and chatbots!",
]
def preprocess(s):
    toks = word_tokenize(s.lower())
    return [t for t in toks if t not in string.punctuation and t not in stop_words]

cleaned = [preprocess(d) for d in docs]
vocab = sorted({w for d in cleaned for w in d})
df = pd.DataFrame([[Counter(d)[w] for w in vocab] for d in cleaned], columns=vocab)
df.insert(0, "original_text", docs)
df.to_csv("preprocessed_dataset.csv", index=False)
df
```
