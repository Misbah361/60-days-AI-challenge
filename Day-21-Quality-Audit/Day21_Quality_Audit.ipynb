```python
DOCUMENTS = []
for i, text, src, cat, dtype, date in raw:
    DOCUMENTS.append({
        "id": f"doc{i}",
        "title": src,            # e.g. policy_2024.pdf, shown as the citation
        "category": cat,
        "document_type": dtype,
        "date": date,            # doc 20 will be None, that's the edge case
        "text": text,
    })

print("Number of documents:", len(DOCUMENTS))
print(DOCUMENTS[1])
```

```python
import faiss
from sentence_transformers import SentenceTransformer
from langchain_text_splitters import RecursiveCharacterTextSplitter

embedder = SentenceTransformer("all-MiniLM-L6-v2")
splitter = RecursiveCharacterTextSplitter(chunk_size=300, chunk_overlap=50)

chunks = []
for doc in DOCUMENTS:
    for i, piece in enumerate(splitter.split_text(doc["text"])):
        chunks.append({
            "chunk_id": f'{doc["id"]}-{i}',
            "title": doc["title"],
            "category": doc["category"],
            "document_type": doc["document_type"],
            "date": doc["date"],
            "text": piece,
        })

vectors = embedder.encode([c["text"] for c in chunks],
                          normalize_embeddings=True).astype("float32")
index = faiss.IndexFlatIP(vectors.shape[1])
index.add(vectors)

print("Number of chunks:", len(chunks))
print("First chunk:", chunks[0])
```

```python
def retrieve(query, top_k=3, category=None, document_type=None, min_date=None):
    # 1. Turn the question into numbers (same model as the chunks)
    q = embedder.encode([query], normalize_embeddings=True).astype("float32")

    # 2. Search ALL chunks, so filtering can't leave us with too few results
    scores, idxs = index.search(q, len(chunks))

    # 3. Walk through results from best to worst, keep only those that pass the filters
    results = []
    for score, i in zip(scores[0], idxs[0]):
        c = chunks[i]
        if category and c["category"] != category:
            continue
        if document_type and c["document_type"] != document_type:
            continue
        if min_date and (c["date"] is None or c["date"] < min_date):
            continue
        results.append({**c, "score": float(score)})
        if len(results) == top_k:
            break
    return results
```

```python
import torch
from transformers import pipeline

llm = pipeline("text-generation",
               model="Qwen/Qwen2.5-1.5B-Instruct",
               torch_dtype=torch.float16,
               device_map="auto")
print("Model loaded")
```

```python
CONFIDENCE_THRESHOLD = 0.3

GROUNDING_PROMPT = """You are a knowledge assistant for Veridia Labs.
Answer ONLY using the context below. Never use outside knowledge.
Each context item shows [chunk_id | source file | date].
If sources conflict, trust the one with the most recent date.
If the context does not contain the answer, reply exactly:
"I don't have enough information in my knowledge base to answer that."
Mention the chunk_id(s) you used in your answer.

Context:
{context}

Question: {question}
Answer:"""


def ask(query, top_k=3, **filters):
    hits = retrieve(query, top_k=top_k, **filters)
    top_score = max((h["score"] for h in hits), default=0.0)

    # newest source first, so the model sees the current info before the old
    ordered = sorted(hits, key=lambda h: h["date"] or "", reverse=True)
    context = "\n".join(
        f'[{h["chunk_id"]} | {h["title"]} | {h["date"]}] {h["text"]}'
        for h in ordered)

    prompt = GROUNDING_PROMPT.format(context=context, question=query)
    out = llm([{"role": "user", "content": prompt}],
              max_new_tokens=150, do_sample=False)
    answer = out[0]["generated_text"][-1]["content"].strip()

    low_conf = top_score < CONFIDENCE_THRESHOLD
    if low_conf:
        answer += ("\n\n⚠️ Low confidence: the best retrieval match was weak "
                   "(score below 0.3), so this answer may not be well-supported.")

    return {
        "answer": answer,
        "sources": [{"chunk_id": h["chunk_id"], "title": h["title"],
                     "score": round(h["score"], 3)} for h in hits],
        "top_score": round(top_score, 3),
        "low_confidence": low_conf,
    }
```

```python
import re

def newest_versions_only(hits):
    # policy_2021.pdf and policy_2024.pdf -> same base name "policy.pdf"
    base = lambda t: re.sub(r"_(19|20)\d{2}(?=\.)", "", t)
    best = {}
    for h in hits:
        k = base(h["title"])
        if k not in best or (h["date"] or "") > (best[k]["date"] or ""):
            best[k] = h
    return [h for h in hits if best[base(h["title"])] is h]


def ask(query, top_k=3, **filters):
    # fetch extra, drop outdated versions, then keep the best top_k
    hits = retrieve(query, top_k=6, **filters)
    hits = newest_versions_only(hits)[:top_k]
    top_score = max((h["score"] for h in hits), default=0.0)
    low_conf = top_score < CONFIDENCE_THRESHOLD

    if low_conf:
        # gate: don't let the model answer from its own knowledge
        answer = REFUSAL
    else:
        ordered = sorted(hits, key=lambda h: h["date"] or "", reverse=True)
        context = "\n".join(
            f'[{h["chunk_id"]} | {h["title"]} | {h["date"]}] {h["text"]}'
            for h in ordered)
        prompt = GROUNDING_PROMPT.format(refusal=REFUSAL, context=context, question=query)
        out = llm([{"role": "user", "content": prompt}],
                  max_new_tokens=150, do_sample=False)
        answer = out[0]["generated_text"][-1]["content"].strip()
        if not answer.startswith(REFUSAL[:30]):
            cites = ", ".join(f'{h["chunk_id"]} ({h["title"]})' for h in hits)
            answer += f"\n\nSources: {cites}"

    if low_conf:
        answer += ("\n\n⚠️ Low confidence: the best retrieval match was weak "
                   "(score below 0.3), so this answer may not be well-supported.")

    return {
        "answer": answer,
        "sources": [{"chunk_id": h["chunk_id"], "title": h["title"],
                     "score": round(h["score"], 3)} for h in hits],
        "top_score": round(top_score, 3),
        "low_confidence": low_conf,
    }
```

```python
import time

def ask_timed(query, top_k=3):
    t = {}

    # 1. embed the question
    s = time.perf_counter()
    q = embedder.encode([query], normalize_embeddings=True).astype("float32")
    t["embed"] = time.perf_counter() - s

    # 2. FAISS search
    s = time.perf_counter()
    scores, idxs = index.search(q, len(chunks))
    hits = [{**chunks[i], "score": float(sc)} for sc, i in zip(scores[0][:6], idxs[0][:6])]
    t["search"] = time.perf_counter() - s

    # 3. drop outdated versions
    s = time.perf_counter()
    hits = newest_versions_only(hits)[:top_k]
    t["version_filter"] = time.perf_counter() - s

    top_score = max((h["score"] for h in hits), default=0.0)
    low_conf = top_score < CONFIDENCE_THRESHOLD

    # 4. LLM generation (skipped by the confidence gate)
    s = time.perf_counter()
    if low_conf:
        answer = REFUSAL
    else:
        ordered = sorted(hits, key=lambda h: h["date"] or "", reverse=True)
        context = "\n".join(f'[{h["chunk_id"]} | {h["title"]} | {h["date"]}] {h["text"]}' for h in ordered)
        prompt = GROUNDING_PROMPT.format(refusal=REFUSAL, context=context, question=query)
        out = llm([{"role": "user", "content": prompt}], max_new_tokens=150, do_sample=False)
        answer = out[0]["generated_text"][-1]["content"].strip()
    t["llm"] = time.perf_counter() - s

    if not low_conf and not answer.startswith(REFUSAL[:30]):
        answer += "\n\nSources: " + ", ".join(f'{h["chunk_id"]} ({h["title"]})' for h in hits)
    if low_conf:
        answer += "\n\n⚠️ Low confidence: the best retrieval match was weak (score below 0.3), so this answer may not be well-supported."

    t["total"] = sum(t.values())
    return {"answer": answer,
            "sources": [{"chunk_id": h["chunk_id"], "title": h["title"], "score": round(h["score"], 3)} for h in hits],
            "top_score": round(top_score, 3), "low_confidence": low_conf, "timing": t}

# warm-up: the very first call is always slower (GPU and caches), so we time it separately
r = ask_timed("How long do I have to ask for a refund?")
print("FIRST (cold) call:", {k: round(v, 3) for k, v in r["timing"].items()})
r = ask_timed("How long do I have to ask for a refund?")
print("SECOND (warm) call:", {k: round(v, 3) for k, v in r["timing"].items()})
```

```python
import pandas as pd
RESULTS = {}   # we collect every dimension's results here for the final report

def body(answer):
    return answer.split("\n\nSources:")[0].split("\n\n⚠️")[0].strip()

def check(answer, expected, forbidden=()):
    b = body(answer).lower()
    if expected == ["REFUSE"]:
        return b.startswith("i don't have enough information")
    ok = all(any(alt.lower() in b for alt in e.split("|")) for e in expected)
    bad = any(f.lower() in b for f in forbidden)
    return ok and not bad

def run_suite(name, cases):
    rows = []
    for q, expected, forbidden in cases:
        try:
            r = ask_timed(q)
            passed = check(r["answer"], expected, forbidden)
            ans = body(r["answer"])
        except Exception as e:
            passed, ans = False, f"CRASH: {e}"
        rows.append({"query": q, "pass": passed, "answer": ans})
    df = pd.DataFrame(rows)
    RESULTS[name] = df
    print(f"{name.upper()}: {df['pass'].sum()}/{len(df)} passed = {df['pass'].mean()*100:.0f}%\n")
    for _, row in df.iterrows():
        print(("PASS" if row["pass"] else "FAIL"), "|", row["query"])
        print("      ", row["answer"][:160].replace("\n", " "))
    return df

# ---- ACCURACY: correct facts from the CURRENT source (old versions are forbidden) ----
accuracy_cases = [
    ("What is the refund window?",                          ["30"],                  ["14 days"]),
    ("How much does the Starter plan cost?",                ["$19"],                 ["$9 "]),
    ("Which encryption standard protects stored data?",     ["AES-256"],             ["AES-128"]),
    ("How many requests per minute can Enterprise make?",   ["5000"],                []),
    ("Do Enterprise customers get phone support?",          ["24/7|phone"],          []),
    ("How many days a week can staff work remotely?",       ["3"],                   ["5 days per week"]),
    ("What did release 4.2 add?",                           ["SAML", "dark mode"],   []),
    ("What did release 3.0 launch?",                        ["analytics", "CSV"],    []),
    ("Is TLS 1.2 still used for data in transit?",          ["1.3"],                 []),
    ("How long until deleted data leaves the backups?",     ["90"],                  []),
]
acc = run_suite("accuracy", accuracy_cases)

# ---- RELIABILITY: same facts under typos, shouting, fragments, emoji, empty input ----
reliability_cases = [
    ("refnd policy how many dayz",                          ["30"],                  ["14 days"]),
    ("HOW LONG TO GET MY MONEY BACK???",                    ["30|5 business"],       []),
    ("refund",                                              ["30"],                  ["14 days"]),
    ("can i work from home",                                ["3"],                   ["5 days per week"]),
    ("pro plan rate limit",                                 ["1000"],                ["100 requests"]),
    ("tell me how veridia protects data at rest encryption",["AES-256"],             ["AES-128"]),
    ("support hours?",                                      ["9am"],                 ["10am"]),
    ("   how much is pro   ",                               ["49"],                  ["$29"]),
    ("refund window 😀",                                    ["30"],                  ["14 days"]),
    ("",                                                    ["REFUSE"],              []),
]
rel = run_suite("reliability", reliability_cases)
```

```python
latency_queries = [
    "What is the refund window?", "How much does the Pro plan cost?",
    "Which encryption protects stored data?", "How many requests per minute on Enterprise?",
    "Do Enterprise customers get phone support?", "How many days can staff work remotely?",
    "What did release 4.2 add?", "How long until deleted data leaves backups?",
    "What TLS version is used in transit?", "When are the support hours?",
]
ask_timed("warm up")   # discard the first call so GPU start-up doesn't skew the average

timings = pd.DataFrame([ask_timed(q)["timing"] for q in latency_queries])
avg = timings.mean()
share = (avg / avg["total"] * 100).round(1)

report = pd.DataFrame({"avg_seconds": avg.round(4), "share_%": share})
print(report)
print(f"\nAverage end-to-end latency over {len(latency_queries)} queries: {avg['total']:.3f} s")
print(f"Slowest query: {timings['total'].max():.3f} s | fastest: {timings['total'].min():.3f} s")
print("Slowest component:", avg.drop("total").idxmax())
RESULTS["latency"] = report
```

```python
absent_queries = [
    "What is the weather like in Srinagar today?",
    "Who won the last cricket World Cup?",
    "How do I bake a chocolate cake?",
    "Explain how photosynthesis works.",
    "What is Veridia Labs' parental leave policy?",    # trap: plausible but not in the KB
]

rows = []
for q in absent_queries:
    r = ask_timed(q)
    refused = body(r["answer"]).lower().startswith("i don't have enough information")
    rows.append({"query": q, "refused": refused, "top_score": r["top_score"],
                 "caught_by": "gate (score < 0.3)" if r["low_confidence"] else ("grounding prompt" if refused else "NOT CAUGHT"),
                 "answer": body(r["answer"])})
deg = pd.DataFrame(rows)
RESULTS["degradation"] = deg

print(f"GRACEFUL DEGRADATION: {deg['refused'].sum()}/{len(deg)} refused honestly\n")
for _, row in deg.iterrows():
    print("REFUSED" if row["refused"] else "ANSWERED ANYWAY", "|", row["query"])
    print(f"      top_score {row['top_score']} | caught by: {row['caught_by']}")
    print("      ", row["answer"][:200].replace("\n", " "))
```

```python
import re

# (question, chunk id that actually contains the answer)
transparency_cases = [
    ("What is the refund window?",                        "doc1-0"),
    ("How much does the Starter plan cost?",              "doc3-0"),
    ("Which encryption standard protects stored data?",   "doc5-0"),
    ("How many requests per minute can Enterprise make?", "doc12-0"),
    ("Do Enterprise customers get phone support?",        "doc9-0"),
    ("How many days a week can staff work remotely?",     "doc17-0"),
    ("What did release 4.2 add?",                         "doc15-0"),
    ("How long until deleted data leaves the backups?",   "doc8-0"),
    ("What TLS version is used for data in transit?",     "doc5-0"),
    ("How long does a refund take to reach my card?",     "doc20-0"),
]

rows = []
for q, right in transparency_cases:
    r = ask_timed(q)
    src_line = r["answer"].split("Sources:")[-1] if "Sources:" in r["answer"] else ""
    ids = [s["chunk_id"] for s in r["sources"]]
    rows.append({
        "query": q,
        "has_citation": "Sources:" in r["answer"],
        "right_source_listed": right in ids,
        "readable": not re.search(r"doc\d+-\d+", src_line),     # chunk IDs are jargon
        "sources_listed": len(ids),
        "has_date": bool(re.search(r"\d{4}", src_line)),
    })
tr = pd.DataFrame(rows)
tr["transparent"] = tr.has_citation & tr.right_source_listed & tr.readable
RESULTS["transparency"] = tr

print(tr[["query", "has_citation", "right_source_listed", "readable", "sources_listed"]])
print(f"\nHas citation:        {tr.has_citation.mean()*100:.0f}%")
print(f"Right source listed: {tr.right_source_listed.mean()*100:.0f}%")
print(f"Readable to a user:  {tr.readable.mean()*100:.0f}%")
print(f"Fully transparent:   {tr.transparent.sum()}/{len(tr)}")

print("\n--- RAW answer for the trap question (is a Sources line attached to a 'no info' answer?) ---")
print(ask_timed("What is Veridia Labs' parental leave policy?")["answer"])
```

```python
user_style = [
    # (query, expected fact for a rough auto-check, or None = judge by reading)
    ("hey so i bought the annual thing last week and changed my mind, can i get my money back or nah?", ["pro-rata|30"]),
    ("how much is it", None),                                            # vague: which plan?
    ("Hi there, I'm trying to understand whether or not our team members are allowed to do their work from home at all, and if so how often that would be permitted per week?", ["3"]),
    ("is my stuff safe", ["AES-256|encrypt"]),                           # vague
    ("my api keeps giving me an error when i send lots of requests", ["429|rate limit"]),
    ("do u guys have a phone number i can call at midnight", ["24/7|enterprise"]),
    ("whats new in the latest version", ["4.2|SAML"]),
    ("can i pay by card and is it secure", ["token"]),
    ("pricing pls", ["19|49"]),
    ("i deleted my account yesterday, when is my data actually gone?", ["90"]),
]

rows = []
for q, expected in user_style:
    r = ask_timed(q)
    ans = body(r["answer"])
    auto = None if expected is None else any(e.lower() in ans.lower() for e in expected[0].split("|"))
    rows.append({"query": q, "answer": ans, "top_score": r["top_score"],
                 "refused": ans.lower().startswith("i don't have enough"),
                 "auto_check": auto, "sources": [s["chunk_id"] for s in r["sources"]]})
us = pd.DataFrame(rows)
RESULTS["user_style"] = us

for i, row in us.iterrows():
    tag = {True: "AUTO-PASS", False: "AUTO-FAIL", None: "JUDGE BY READING"}[row["auto_check"]]
    print(f"\n[{i+1}] {tag} | score {row['top_score']} | refused: {row['refused']}")
    print("   USER:  ", row["query"])
    print("   SYSTEM:", row["answer"][:260].replace("\n", " "))
```
